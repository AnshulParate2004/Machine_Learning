## Identify and handle missing values

### Subtask:
Identify missing values in the dataset and apply appropriate techniques to handle them.


**Reasoning**:
Calculate and display the number of missing values per column to identify columns with missing data.



```python
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values)
```

**Reasoning**:
Handle missing values in the 'Age', 'Cabin', and 'Embarked' columns based on the extent of missingness and the nature of the data. For 'Age', impute with the median. For 'Cabin', which has a high percentage of missing values, drop the column. For 'Embarked', impute with the mode as it is a categorical variable with few missing values.



```python
df['Age'].fillna(df['Age'].median(), inplace=True)
df.drop('Cabin', axis=1, inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
```

**Reasoning**:
Verify that all missing values have been handled by calculating and displaying the number of missing values for each column again.



```python
missing_values_after = df.isnull().sum()
print("Missing values per column after handling:")
print(missing_values_after)
```

### Subtask:

**Reasoning**:

## Identify and handle outliers

### Subtask:
Identify outliers in numerical columns and handle them using appropriate methods.


**Reasoning**:
Select numerical columns, calculate IQR, define outlier bounds, identify outliers, and handle them by capping within a single code block.



```python
import numpy as np

numerical_cols = df.select_dtypes(include=np.number).columns

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    if not outliers.empty:
        df[col] = np.where(df[col] < lower_bound, lower_bound, df[col])
        df[col] = np.where(df[col] > upper_bound, upper_bound, df[col])
        print(f"Outliers in column '{col}' handled by capping.")
    else:
        print(f"No outliers found in column '{col}'.")
```
