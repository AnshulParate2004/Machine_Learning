```python
from bs4 import BeautifulSoup

with open('smartprix.html' , 'r' , encoding='utf-8') as f:
    html = f.read()

soup = BeautifulSoup(html , 'lxml')

print(soup.prettify())
```

```python
mobile_names = [tag.get_text(strip=True) for tag in soup.find_all('h2')]
```

```python
mobile_price = [price.get_text(strip=True) for price in soup.find_all('span' , class_='price')]
```

```python
mobile_img = [img.get('src') for img in soup.find_all('img' , class_='sm-img')]
```

```python
rating = [rating for rating in soup.find_all('span' , class_='sm-rating')]
```

```python
print(len(mobile_names))
print(len(mobile_price))
print(len(mobile_img))
print(len(rating))
```

```python
mobiles = [prod for prod in soup.find_all('div' , class_='sm-product has-tag has-features has-actions')]
```

```python
mobile_names = []
mobile_prices = []
mobile_specs_scores = []
mobile_imgs = []
mobile_ratings = []
mobile_sims = []
mobile_processors = []
mobile_ram = []
mobile_battery = []
mobile_display_sizes = []
mobile_cameras = []
mobile_android_version = []
```

```python
import re
import numpy as np

for mobile in mobiles:
    mobile_name = mobile.find('h2').text
    mobile_names.append(mobile_name)

    mobile_price = mobile.find('span' , class_='price').text
    mobile_prices.append(mobile_price)

    mobile_specs_score = mobile.find('b').text
    mobile_specs_scores.append(mobile_specs_score)

    mobile_img = mobile.find('div' , class_='sm-img-wrap').find('img').get('src')
    mobile_imgs.append(mobile_img)

    mobile_rating = mobile.find('div' , class_='rating').find('span' , class_='sm-rating').get('style')
    rating = re.search(r'--rating:\s*([\d.]+)', mobile_rating)

    if rating:
      rating = float(rating.group(1))
      mobile_ratings.append(rating)
    else:
      mobile_ratings.append(np.nan)

    li_items = mobile.find('ul' , class_='sm-feat specs').find_all('li')

    try:
        sim = li_items[0].text.strip()
    except:
        sim = np.nan
    mobile_sims.append(sim)

    try:
        processors = li_items[1].text.strip()
    except:
        processors = np.nan
    mobile_processors.append(processors)

    try:
        ram = li_items[2].text.strip()
    except:
        ram = np.nan
    mobile_ram.append(ram)

    try:
        battery = li_items[3].text.strip()
    except:
        battery = np.nan
    mobile_battery.append(battery)

    try:
        display_sizes = li_items[4].text.strip()
    except:
        display_sizes = np.nan
    mobile_display_sizes.append(display_sizes)

    try:
        cameras = li_items[5].text.strip()
    except:
        cameras = np.nan
    mobile_cameras.append(cameras)

    try:
        android_version = li_items[6].text.strip()
    except:
        android_version = np.nan
    mobile_android_version.append(android_version)
```

```python
import pandas as pd

df = pd.DataFrame({
    'Name': mobile_names,
    'Price': mobile_prices,
    'Spec Score': mobile_specs_scores,
    'Image URL': mobile_imgs,
    'Rating': mobile_ratings,
    'Sim': mobile_sims,
    'Processor': mobile_processors,
    'RAM': mobile_ram,
    'Battery': mobile_battery,
    'Display Size': mobile_display_sizes,
    'Camera': mobile_cameras,
    'Android Version': mobile_android_version
})
```
