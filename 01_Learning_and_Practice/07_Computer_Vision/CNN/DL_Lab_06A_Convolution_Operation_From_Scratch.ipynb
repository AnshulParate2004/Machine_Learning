```python
import numpy as np

# Input Image (4x4)
X = np.array([
    [1, 0, 2, 3],
    [4, 6, 6, 8],
    [3, 1, 1, 0],
    [1, 2, 2, 4]
])

# Kernel (3x3)
K = np.array([
    [1, 0, -1],
    [1, -1, 0],
    [0, 1, -1]
])
```

```python
def convolution(X, K):
    n = X.shape[0]
    f = K.shape[0]
    output_size = n - f + 1

    output = np.zeros((output_size, output_size))

    for i in range(output_size):
        for j in range(output_size):
            region = X[i:i+f, j:j+f]
            output[i, j] = np.sum(region * K)

    return output

conv_output = convolution(X, K)
print("Convolution Output:\n", conv_output)
```

```python
def relu(x):
    return np.maximum(0, x)

activated_output = relu(conv_output)
print("After ReLU:\n", activated_output)
```

```python
flattened = activated_output.flatten()
print("Flattened Output:", flattened)
```

```python
# Initialize weights and bias
np.random.seed(0)
W = np.random.randn(flattened.shape[0])
b = np.random.randn()

# Dense forward
dense_output = np.dot(flattened, W) + b
print("Dense Output:", dense_output)
```

```python
target = 5

loss = (target - dense_output) ** 2
print("Loss (MSE):", loss)
```
